```python
import json
from pathlib import Path

BASE_DIR = Path().resolve().parents[1]

with open(BASE_DIR / "data" / "rule_metadata_merged.json", encoding="utf-8") as f:
    rule_data = json.load(f)

rules = rule_data["unique_rules"]

# query-pos 쌍 생성
query_pos_pairs = []

for rule in rules:
    description = rule["description"]
    
    # 대표 질문
    if rule.get("question"):
        query_pos_pairs.append({
            "query": rule["question"],
            "pos":   description,
            "rule_title": rule["title"]
        })
    
    # merged_questions
    for q in rule.get("merged_questions", []):
        query_pos_pairs.append({
            "query": q,
            "pos":   description,
            "rule_title": rule["title"]
        })

print(f"총 query-pos 쌍: {len(query_pos_pairs)}개")

# 샘플 확인
for pair in query_pos_pairs[:3]:
    print(f"\nquery: {pair['query']}")
    print(f"pos:   {pair['pos']}")
    print(f"title: {pair['rule_title']}")
```

```python
# question + merged_questions 전부 포함
query_pos_all = []
for rule in rules:
    description = rule["description"]
    title = rule["title"]
    
    if rule.get("question"):
        query_pos_all.append({
            "query": rule["question"],
            "pos":   description,
            "title": title
        })
    
    for q in rule.get("merged_questions", []):
        query_pos_all.append({
            "query": q,
            "pos":   description,
            "title": title
        })

print(f"총 query-pos 쌍: {len(query_pos_all)}개")

# 저장
with open(BASE_DIR / "data" / "finetune_embd_query_pos.json", "w", encoding="utf-8") as f:
    json.dump(query_pos_all, f, ensure_ascii=False, indent=2)

print(f"✅ 저장 완료 → data/finetune_embd_query_pos.json")
```
