```python
import json
from collections import Counter

with open("../data/rule_metadata_merged.json", "r", encoding="utf-8") as f:
    rules = json.load(f)

# rule당 현재 질문 수
shortage = []
for r in rules["unique_rules"]:
    count = len(r.get("merged_questions", [])) + 1
    if count < 10:
        shortage.append({
            "title": r["title"],
            "description": r["description"],
            "current": count,
            "need": 10 - count
        })

print(f"10개 미만 rule: {len(shortage)}개")
total_need = sum(s["need"] for s in shortage)
print(f"추가 생성 필요 질문 수: {total_need}개")
```

```python
from collections import Counter

with open("../data/rule_metadata_merged.json", "r", encoding="utf-8") as f:
    rules = json.load(f)

titles = [r["title"] for r in rules["unique_rules"]]
title_count = Counter(titles)

duplicates = {t: c for t, c in title_count.items() if c > 1}
print(f"중복 title 수: {len(duplicates)}개")
print(f"실제 unique title: {len(set(titles))}개")
print(f"전체 rule 수: {len(titles)}개")
print(f"\n중복 title:")
for t, c in sorted(duplicates.items(), key=lambda x: -x[1]):
    print(f"  {t}: {c}개")
```

```python
dups = [r for r in rules["unique_rules"] if r["title"] == "애견동반"]
for r in dups:
    print(f"description: {r['description']}")
    print(f"keywords 수: {len(r['keywords'])}")
    print()
```

```python
# title 기준으로 그룹핑
groups = defaultdict(list)
for r in rules["unique_rules"]:
    groups[r["title"]].append(r)

# 합치기
merged = []
for title, group in groups.items():
    if len(group) == 1:
        merged.append(group[0])
    else:
        all_keywords = []
        seen = set()
        for r in group:
            for kw in r["keywords"]:
                if kw not in seen:
                    all_keywords.append(kw)
                    seen.add(kw)

        all_questions = []
        seen_q = set()
        for r in group:
            for q in r.get("merged_questions", []):
                if q not in seen_q:
                    all_questions.append(q)
                    seen_q.add(q)

        merged.append({
            "title": title,
            "description": max(group, key=lambda x: len(x["description"]))["description"],
            "keywords": all_keywords,
            "question": group[0]["question"],
            "merged_questions": all_questions,
            "confidence": group[0]["confidence"],
            "reason": group[0]["reason"],
        })

print(f"합치기 전: {len(rules['unique_rules'])}개")
print(f"합치기 후: {len(merged)}개")

# 저장
output = {"unique_rules": merged}
with open("../data/rule_metadata_merged.json", "w", encoding="utf-8") as f:
    json.dump(output, f, ensure_ascii=False, indent=2)

print("저장 완료!")
```
