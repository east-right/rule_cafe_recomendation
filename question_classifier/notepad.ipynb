```python
import json

with open("data/trainval.jsonl", "r", encoding="utf-8") as f:
    trainval = [json.loads(line) for line in f]

with open("data/test.jsonl", "r", encoding="utf-8") as f:
    test = [json.loads(line) for line in f]

# 질문 완전 중복 확인
trainval_questions = set(d["question"] for d in trainval)
test_questions = set(d["question"] for d in test)

overlap = trainval_questions & test_questions
print(f"완전 중복 질문: {len(overlap)}개")
for q in list(overlap)[:5]:
    print(f"  {q}")
```

```python
import json
from collections import Counter

with open("./data/trainval.jsonl", "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

ID2LABEL = {0: "non-menu", 1: "menu-only", 2: "menu-complex", 3: "invalid"}

for label, name in ID2LABEL.items():
    samples = [d for d in data if d["label"] == label]
    print(f"\n[{name}] 총 {len(samples)}개 - 샘플 10개:")
    for d in samples[:10]:
        print(f"  {d['question']}")
```

```python
import json

with open("./data/augmented.jsonl", "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

# non-menu인데 메뉴 키워드 포함된 것들
menu_words = ["아아", "뜨아", "아바라", "라떼", "아메리카노", "케이크", "빵", "주스", "커피"]
suspicious = [
    d for d in data 
    if d["label"] == 0  # non-menu
    and any(w in d["question"] for w in menu_words)
]
print(f"non-menu인데 메뉴 키워드 포함: {len(suspicious)}개")
for d in suspicious[:5]:
    print(f"  {d['question']}")
```

```python
actually_wrong = [d for d in suspicious if "아아" in d["question"] or "뜨아" in d["question"] or "아바라" in d["question"]]
print(f"실제로 잘못된 케이스: {len(actually_wrong)}개")
for d in actually_wrong:
    print(f"  {d['question']}")
```

```python
import json

with open("./data/augmented.jsonl", "r", encoding="utf-8") as f:
    data = [json.loads(line) for line in f]

# 제거할 질문들
remove_questions = {
    "아아 마시면서 편하게 수다 떨 수 있는 카페 없나?",
    "디저트 맛집 카페 뭐 있는지 아아? 추천 좀~",
    "신림역 부근에 분위기 짱이고 넓은 카페 아아 마실 겸 카공하기 좋은 곳 있음?",
    "주차 잘 되고 편한 카페 ㄱㄱ? 여기 근처에 뜨아 마시기 좋은 곳 없나?",
    "아아 맛있는 카페 어디임? 찐 추천 좀!",
}

cleaned = [d for d in data if d["question"] not in remove_questions]
print(f"제거 전: {len(data)}개 → 제거 후: {len(cleaned)}개")

with open("./data/augmented.jsonl", "w", encoding="utf-8") as f:
    for d in cleaned:
        f.write(json.dumps(d, ensure_ascii=False) + "\n")

print("완료!")
```
